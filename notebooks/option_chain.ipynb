```python
def generate_options_surface(options_df, ohlcv_df, stock, date, splits_file="data/raw/stocks/csv/split.csv", option_type="Put", show_surface=True):
    """
    Generate volatility surface from options chain data with split adjustments.
    
    Parameters:
    -----------
    options_df : pl.DataFrame
        DataFrame containing options data
    ohlcv_df : pl.DataFrame
        DataFrame containing OHLCV data (already split-adjusted)
    stock : str
        Stock symbol
    date : str or datetime.date or pl.Date
        Date for visualization
    splits_file : str
        Path to the splits CSV file
    option_type : str
        "Put", "Call", or "Both" (default: "Put")
    show_surface : bool
        Whether to display the surface (default: True)
    
    Returns:
    --------
    tuple: (fig, surface_dict)
        fig: plotly figure object
        surface_dict: dictionary with surface data
    """
    # Ensure date is in pl.Date format
    if isinstance(date, str):
        y, m, d = map(int, date.split("-"))
        date = pl.date(y, m, d)
    elif isinstance(date, datetime.date):
        date = pl.date(date.year, date.month, date.day)
    
    # Find nearest options date not after requested date
    available_dates = options_df.filter(
        (pl.col("act_symbol") == stock) & 
        (pl.col("date") <= date)
    ).select("date").unique().sort("date", descending=True)
    
    if available_dates.height == 0:
        raise ValueError(f"No options data found for {stock} on or before {date}")
    
    options_date = available_dates[0, 0]
    
    # Find nearest OHLCV date not after requested date
    available_ohlcv_dates = ohlcv_df.filter(
        (pl.col("act_symbol") == stock) & 
        (pl.col("date") <= date)
    ).select("date").unique().sort("date", descending=True)
    
    if available_ohlcv_dates.height == 0:
        raise ValueError(f"No OHLCV data found for {stock} on or before {date}")
    
    ohlcv_date = available_ohlcv_dates[0, 0]
    
    # Get underlying price for moneyness calculation
    spot_price = ohlcv_df.filter(
        (pl.col("act_symbol") == stock) & 
        (pl.col("date") == ohlcv_date)
    ).select("close")[0, 0]
    
    # Load split data for the specific stock
    print(f"Loading split data for {stock}...")
    try:
        splits = (
            pl.read_csv(splits_file)
            .filter(pl.col("act_symbol") == stock)
            .with_columns(
                pl.col("ex_date").str.to_date("%Y-%m-%d"),
                pl.when((pl.col("to_factor") == 0) | (pl.col("for_factor") == 0))
                  .then(1.0)
                  .otherwise(pl.col("to_factor") / pl.col("for_factor"))
                  .alias("split_factor")
            )
            .filter(pl.col("split_factor") != 1.0)
            .select(["act_symbol", "ex_date", "split_factor"])
        )
        
        # Calculate adjustment factor for options data
        relevant_splits = splits.filter(
            (pl.col("ex_date") > options_date)
        )
        
        # Calculate cumulative split factor
        if relevant_splits.height > 0:
            cumulative_split_factor = relevant_splits["split_factor"].product()
            print(f"Found {relevant_splits.height} split(s) since {options_date}. " +
                  f"Cumulative factor: {cumulative_split_factor}")
        else:
            cumulative_split_factor = 1.0
            print("No splits found between option date and target date.")
    except Exception as e:
        print(f"Error processing splits: {e}. Assuming no splits.")
        cumulative_split_factor = 1.0
    
    # Filter options data for the selected date
    filtered_options = options_df.filter(
        (pl.col("act_symbol") == stock) & 
        (pl.col("date") == options_date)
    )
    
    # Filter by option type if specified
    if option_type != "Both":
        filtered_options = filtered_options.filter(pl.col("call_put") == option_type)
    
    # Get available expirations
    expirations = filtered_options.select("expiration").unique().sort("expiration")
    
    if expirations.height == 0:
        raise ValueError(f"No {option_type} expiration data found for {stock} on {options_date}")
    
    # Calculate days to expiration for each expiration date
    expirations_with_days = []
    for exp_date in expirations["expiration"]:
        # Convert to Python date objects for proper subtraction
        if hasattr(options_date, 'year'):
            # Polars date object
            options_date_py = datetime.date(
                options_date.year, options_date.month, options_date.day
            )
            exp_date_py = datetime.date(
                exp_date.year, exp_date.month, exp_date.day
            )
        else:
            # Handle case where it might be a Python date already
            options_date_py = options_date
            exp_date_py = exp_date
            
        days = (exp_date_py - options_date_py).days
        expirations_with_days.append((exp_date, days))
    
    # Filter expirations to have a reasonable range (5 to 180 days)
    expirations_with_days = [(exp, days) for exp, days in expirations_with_days if 5 <= days <= 180]
    
    if not expirations_with_days:
        raise ValueError(f"No suitable expirations found for {stock} on {options_date}")
    
    # Prepare data for the surface
    surface_data = []
    
    for expiration, days_to_exp in expirations_with_days:
        # Get options for this expiration
        options = filtered_options.filter(pl.col("expiration") == expiration)
        
        # Process each option
        for row in options.iter_rows(named=True):
            strike = row["strike"]
            implied_vol = row["vol"]  # IV is stored in the 'vol' column
            
            # Skip if vol is missing or invalid
            if implied_vol is None or implied_vol <= 0 or implied_vol > 5:
                continue
            
            # Apply split adjustment to strike price
            adjusted_strike = strike / cumulative_split_factor
            
            # Calculate log-moneyness
            log_moneyness = np.log(adjusted_strike / spot_price)
            
            # Store data point
            surface_data.append({
                "days": days_to_exp,
                "moneyness": log_moneyness,
                "vol": implied_vol,
                "strike": adjusted_strike,  # Store adjusted strike
                "original_strike": strike,  # Keep original strike for reference
                "call_put": row["call_put"]
            })
    
    if not surface_data:
        raise ValueError(f"No valid options data found for surface")
    
    # Convert to DataFrame for easier manipulation
    surface_df = pl.DataFrame(surface_data)
    
    # Create moneyness and days grids for the surface
    moneyness = np.linspace(
        surface_df["moneyness"].min() * 0.9,
        surface_df["moneyness"].max() * 1.1,
        50
    )
    
    unique_days = sorted(surface_df["days"].unique().to_list())
    
    K, T = np.meshgrid(moneyness, unique_days)
    vol_surface = np.zeros(K.shape)
    
    # Initialize with NaNs
    vol_surface.fill(np.nan)
    
    # Improved interpolation using SciPy's griddata
    from scipy.interpolate import griddata
    
    # Extract data points for interpolation
    point_days = surface_df["days"].to_numpy()
    point_moneyness = surface_df["moneyness"].to_numpy()
    point_vols = surface_df["vol"].to_numpy()
    
    # Create arrays of grid points
    grid_days = T.flatten()
    grid_moneyness = K.flatten()
    points = np.column_stack([point_days, point_moneyness])
    grid_points = np.column_stack([grid_days, grid_moneyness])
    
    # Perform interpolation (linear is more stable than cubic)
    grid_vols = griddata(points, point_vols, grid_points, method='linear', fill_value=np.nan)
    vol_surface = grid_vols.reshape(T.shape)
    
    # Secondary pass to fill remaining NaNs with nearest neighbor interpolation
    if np.isnan(vol_surface).any():
        valid_mask = ~np.isnan(grid_vols)
        if np.any(valid_mask):  # Only proceed if we have valid points
            nn_vols = griddata(grid_points[valid_mask], grid_vols[valid_mask], 
                              grid_points, method='nearest')
            # Replace NaNs with nearest values
            nan_mask = np.isnan(vol_surface)
            vol_surface_flat = vol_surface.flatten()
            vol_surface_flat[nan_mask.flatten()] = nn_vols[nan_mask.flatten()]
            vol_surface = vol_surface_flat.reshape(T.shape)
    
    # Create 3D surface plot
    fig = go.Figure()
    
    # Add surface (only where we have data)
    # Convert nan values to None for Plotly
    vol_surface_plotly = vol_surface.copy()
    vol_surface_plotly[np.isnan(vol_surface_plotly)] = None
    
    fig.add_trace(go.Surface(
        x=K, 
        y=T, 
        z=vol_surface_plotly, 
        colorscale='Viridis',
        colorbar=dict(title='Implied Volatility'),
        name='IV Surface',
        showscale=True,
        contours={
            "z": {"show": True, "usecolormap": True, "highlightcolor":"white", "project":{"z": True}}
        }
    ))
    
    # Add individual option data points
    fig.add_trace(go.Scatter3d(
        x=point_moneyness,
        y=point_days,
        z=point_vols,
        mode='markers',
        marker=dict(
            size=4, 
            color=point_vols,
            colorscale='Viridis',
            opacity=0.7,
            showscale=False
        ),
        name=f'{option_type} Options'
    ))
    
    # Create split-adjusted title
    split_info = ""
    if cumulative_split_factor != 1.0:
        split_info = f" (Split-adjusted by factor {cumulative_split_factor:.2f})"
    
    # Set axis labels and title
    option_type_str = option_type if option_type != "Both" else "Call/Put"
    fig.update_layout(
        title=f"{option_type_str} Implied Volatility Surface for {stock} on {options_date}{split_info} (Spot: {spot_price:.2f})",
        scene=dict(
            xaxis_title='Log-Moneyness (k)',
            yaxis_title='Days to Expiration',
            zaxis_title='Implied Volatility',
            xaxis=dict(range=[moneyness.min(), moneyness.max()]),
            yaxis=dict(range=[min(unique_days), max(unique_days)]),
            zaxis=dict(range=[0, np.nanmax(vol_surface) * 1.1])
        ),
        width=900,
        height=700,
        margin=dict(r=20, l=10, b=10, t=50)
    )
    
    # Pack surface data
    result_data = {
        'K': K,
        'T': T,
        'vol_surface': vol_surface,
        'moneyness': moneyness,
        'days': unique_days,
        'spot_price': spot_price,
        'options_date': options_date,
        'ohlcv_date': ohlcv_date,
        'raw_data': surface_df,
        'option_type': option_type,
        'split_factor': cumulative_split_factor
    }
    
    # Display if requested
    if show_surface:
        fig.show()
    
    return fig, result_data

# Generate the option volatility surface
options_fig, options_surface = generate_options_surface(
    options_df=option_chain_df,
    ohlcv_df=ohlcv_df,
    stock=stock,
    date=date,
    show_surface=True
)
```

```python
import plotly.graph_objects as go
import numpy as np
import polars as pl
import datetime

def generate_vol_surface(df, stock, date, show_surface=True):
    """
    Generate a volatility surface with proper calendar day time horizons.
    
    Parameters:
    -----------
    df : pl.DataFrame
        DataFrame containing _future columns
    stock : str
        Stock symbol
    date : str or datetime.date or pl.Date or pl.Expr
        Date for visualization
    show_surface : bool
        Whether to display the surface (default: True)
    
    Returns:
    --------
    tuple: (fig, surface_dict)
        fig: plotly figure object
        surface_dict: dictionary with surface data
    """
    # Handle Polars expression as date
    if isinstance(date, pl.Expr):
        print("Date is a Polars expression. Finding an actual date...")
        filtered_df = df.filter((pl.col("act_symbol") == stock) & (pl.col("date") == date))
        if filtered_df.height == 0:
            raise ValueError(f"No data found for stock {stock}")
        date = filtered_df.select("date").row(0)[0]
        print(f"Using date: {date}")
    
    # Filter data for the specific stock and date
    row = df.filter((pl.col("act_symbol") == stock) & (pl.col("date") == date))
    
    if row.height == 0:
        raise ValueError(f"No data found for {stock} on {date}")
    
    # Define trading day windows (these are the windows in your _future columns)
    trading_windows = [10, 15, 20, 25, 30, 35]
    
    # Convert trading days to approximate calendar days
    # A common approximation: calendar days ≈ trading days × (7/5)
    calendar_days = [int(round(window * 7/5)) for window in trading_windows]
    
    # Extract realized values for all timepoints
    realized_data = {}
    for i, trading_window in enumerate(trading_windows):
        days = calendar_days[i]
        years = days / 365.0  # For term structure effects
        
        realized_data[days] = {
            'vol': float(row[f"YZVol_{trading_window}_future"][0]) * np.sqrt(252),  # Annualize
            'moneyness': float(row[f"LogPriceRatio_{trading_window}_future"][0]),
            'skew': float(row[f"VolSkew_{trading_window}_future"][0]),
            'curvature': float(row[f"VolCurvature_{trading_window}_future"][0]),
            'wing_ratio': float(row[f"WingRatio_{trading_window}_future"][0]),
            'mean_reversion': float(row[f"MeanReversion_{trading_window}_future"][0]),
            'vol_of_vol': float(row[f"VolOfVol_{trading_window}_future"][0]),
            'price_vol_corr': float(row[f"PriceVolCorr_{trading_window}_future"][0]),
            'vol_intensity': float(row[f"VolIntensity_{trading_window}_future"][0]),
            'trading_window': trading_window,  # Store original trading days window
            'years': years  # Store this for reference
        }
    
    # Determine moneyness range dynamically
    realized_moneyness = [realized_data[t]['moneyness'] for t in calendar_days]
    min_realized = min(realized_moneyness)
    max_realized = max(realized_moneyness)
    
    # Create wider range around realized moneyness points
    moneyness_padding = 0.15
    moneyness_min = min_realized - moneyness_padding
    moneyness_max = max_realized + moneyness_padding
    moneyness = np.linspace(moneyness_min, moneyness_max, 50)
    
    # Create meshgrid for surface plotting using calendar days
    K, T = np.meshgrid(moneyness, calendar_days)
    vol_surface = np.zeros(K.shape)
    
    # Generate surface with correct mathematics
    for i, days in enumerate(calendar_days):
        data = realized_data[days]
        realized_vol = data['vol']
        realized_k = data['moneyness']
        
        for j, k in enumerate(moneyness):
            # Calculate relative moneyness (distance from realized moneyness)
            relative_k = k - realized_k
            
            # Skew component - asymmetric effect
            skew_effect = 0
            if relative_k < 0:  # Put side (typically higher volatility)
                skew_effect = data['skew'] * abs(relative_k) * data['wing_ratio']
            else:  # Call side
                skew_effect = -data['skew'] * relative_k / max(0.2, data['wing_ratio'])
            
            # Curvature component - creates the smile/frown
            curvature_effect = data['curvature'] * relative_k**2 * (1 + 2.0 * data['vol_of_vol'])
            
            # Price-vol correlation effect
            corr_effect = data['price_vol_corr'] * relative_k * 0.5
            
            # Intensity effect - affects tails
            intensity_effect = (data['vol_intensity'] - 0.5) * 2.0 * abs(relative_k)**2
            
            # Combined effect - must equal 0 when k = realized_k
            combined_effect = skew_effect + curvature_effect + corr_effect + intensity_effect
            
            # Final volatility calculation
            vol = realized_vol * (1.0 + combined_effect)
            
            # Ensure volatility stays within reasonable bounds
            vol = max(0.05, min(vol, 2.0))
            vol_surface[i, j] = vol
    
    # Create 3D surface plot
    fig = go.Figure()
    
    # Add main surface
    fig.add_trace(go.Surface(
        x=K, 
        y=T, 
        z=vol_surface, 
        colorscale='Viridis',
        colorbar=dict(title='Annualized Volatility'),
        name='Volatility Surface'
    ))
    
    # Add actual realized points
    actual_moneyness = [realized_data[t]['moneyness'] for t in calendar_days]
    actual_vols = [realized_data[t]['vol'] for t in calendar_days]
    
    fig.add_trace(go.Scatter3d(
        x=actual_moneyness,
        y=calendar_days,
        z=actual_vols,
        mode='markers',
        marker=dict(size=7, color='red'),
        name='Realized Volatility'
    ))
    
    # Add text annotation showing trading days to calendar days mapping
    annotations = []
    for i, (trade_days, cal_days) in enumerate(zip(trading_windows, calendar_days)):
        annotations.append(
            dict(
                showarrow=False,
                x=min_realized,
                y=cal_days,
                z=0,
                text=f"{trade_days}td",
                xanchor="left",
                font=dict(color="white", size=10)
            )
        )
    
    # Set axis labels and title
    date_str = str(date)
    fig.update_layout(
        title=f"Realized Volatility Surface for {stock} on {date_str}",
        scene=dict(
            xaxis_title='Log-Moneyness (k)',
            yaxis_title='Calendar Days',
            zaxis_title='Annualized Volatility',
            xaxis=dict(range=[moneyness_min, moneyness_max]),
            yaxis=dict(range=[min(calendar_days), max(calendar_days)]),
            zaxis=dict(range=[0, np.nanmax(vol_surface) * 1.1]),
            annotations=annotations
        ),
        width=900,
        height=700,
        margin=dict(r=20, l=10, b=10, t=50)
    )
    
    # Pack surface data using calendar days
    surface_data = {
        'K': K,
        'T': T,
        'vol_surface': vol_surface,
        'calendar_days': calendar_days,
        'trading_windows': trading_windows,
        'moneyness': moneyness,
        'actual_moneyness': actual_moneyness,
        'actual_vols': actual_vols,
        'realized_data': realized_data,
        'date': date_str,
        'stock': stock
    }
    
    # Display if requested
    if show_surface:
        fig.show()
    
    return fig, surface_data

ohlcv_fig, ohlcv_surface = generate_vol_surface(ohlcv_df, stock, date)
```

```python
import plotly.graph_objects as go
import plotly.subplots as sp
import numpy as np
import polars as pl
from scipy.interpolate import griddata

def compare_surfaces(realized_data, options_data, show_plot=True):
    """
    Compare realized volatility surface with options-implied volatility surface,
    properly handling calendar days.
    
    Parameters:
    -----------
    realized_data : dict
        Output from generate_vol_surface function
    options_data : dict
        Output from generate_options_surface function
    show_plot : bool
        Whether to display the comparison plot
        
    Returns:
    --------
    tuple: (fig, comparison_data)
        fig: plotly figure with comparison visualizations
        comparison_data: dictionary with comparison metrics and resampled data
    """
    # Extract data - handle both old and new data structure formats
    realized_K = realized_data['K']
    realized_T = realized_data['T'] 
    realized_vol = realized_data['vol_surface']
    
    options_K = options_data['K']
    options_T = options_data['T']
    options_vol = options_data['vol_surface']
    
    # Get the correct timepoints reference (calendar_days in updated function)
    if 'calendar_days' in realized_data:
        realized_timepoints = realized_data['calendar_days']
    elif 'timepoints' in realized_data:
        realized_timepoints = realized_data['timepoints']
    else:
        # Extract from the grid
        realized_timepoints = realized_T[0, :]
    
    # Create common grid for comparison
    # We'll resample the options surface onto the realized surface grid for consistency
    common_K = realized_K
    common_T = realized_T
    
    # Flatten the options surface for interpolation
    options_days_flat = options_T.flatten()
    options_moneyness_flat = options_K.flatten()
    options_vol_flat = options_vol.flatten()
    
    # Remove NaN values for interpolation
    valid_mask = ~np.isnan(options_vol_flat)
    valid_days = options_days_flat[valid_mask]
    valid_moneyness = options_moneyness_flat[valid_mask]
    valid_vol = options_vol_flat[valid_mask]
    
    if len(valid_vol) == 0:
        raise ValueError("Option surface contains no valid data points for comparison")
    
    # Prepare target points for interpolation
    target_days = common_T.flatten()
    target_moneyness = common_K.flatten()
    target_points = np.column_stack([target_days, target_moneyness])
    
    # Perform interpolation of options surface onto realized grid
    interpolated_options_vol = griddata(
        np.column_stack([valid_days, valid_moneyness]),
        valid_vol,
        target_points,
        method='linear',
        fill_value=np.nan
    ).reshape(common_T.shape)
    
    # Calculate difference: realized - implied (positive = potential overpricing)
    diff_surface = realized_vol - interpolated_options_vol
    
    # Identify common valid areas (where both surfaces have data)
    common_valid = (~np.isnan(realized_vol)) & (~np.isnan(interpolated_options_vol))
    
    # Calculate comparison metrics
    metrics = {}
    if np.any(common_valid):
        realized_vals = realized_vol[common_valid]
        implied_vals = interpolated_options_vol[common_valid]
        diff_vals = diff_surface[common_valid]
        
        metrics = {
            'rmse': np.sqrt(np.mean(np.square(diff_vals))),
            'mae': np.mean(np.abs(diff_vals)),
            'max_diff': np.max(np.abs(diff_vals)),
            'mean_diff': np.mean(diff_vals),
            'median_diff': np.median(diff_vals),
            'std_diff': np.std(diff_vals),
            'correlation': np.corrcoef(realized_vals, implied_vals)[0, 1],
            'common_points': np.sum(common_valid)
        }
    else:
        print("Warning: No common valid points found between surfaces")
        metrics = {
            'rmse': np.nan, 'mae': np.nan, 'max_diff': np.nan,
            'mean_diff': np.nan, 'median_diff': np.nan, 'std_diff': np.nan,
            'correlation': np.nan, 'common_points': 0
        }
    
    # Create multi-panel visualization
    fig = sp.make_subplots(
        rows=2, cols=2,
        specs=[[{'type': 'surface'}, {'type': 'surface'}],
               [{'type': 'surface'}, {'type': 'scatter'}]],
        subplot_titles=[
            'Realized Volatility Surface', 
            'Option-Implied Volatility Surface',
            'Difference (Realized - Implied)',
            'Slice Comparison'
        ],
        horizontal_spacing=0.05,
        vertical_spacing=0.05
    )
    
    # Handle NaN values for plotting
    realized_vol_plot = realized_vol.copy()
    interpolated_options_vol_plot = interpolated_options_vol.copy()
    diff_surface_plot = diff_surface.copy()
    
    realized_vol_plot[np.isnan(realized_vol_plot)] = None
    interpolated_options_vol_plot[np.isnan(interpolated_options_vol_plot)] = None
    diff_surface_plot[np.isnan(diff_surface_plot)] = None
    
    # Determine color scale range for consistency
    vmin = min(
        np.nanmin(realized_vol_plot if realized_vol_plot is not None else np.array([0])),
        np.nanmin(interpolated_options_vol_plot if interpolated_options_vol_plot is not None else np.array([0]))
    )
    vmax = max(
        np.nanmax(realized_vol_plot if realized_vol_plot is not None else np.array([1])),
        np.nanmax(interpolated_options_vol_plot if interpolated_options_vol_plot is not None else np.array([1]))
    )
    
    # 1. Realized Surface
    fig.add_trace(
        go.Surface(
            x=common_K, y=common_T, z=realized_vol_plot,
            colorscale='Viridis',
            cmin=vmin, cmax=vmax,
            name='Realized Volatility',
            showscale=True
        ),
        row=1, col=1
    )
    
    # Add actual data points to realized surface - handle different key names
    if 'actual_moneyness' in realized_data and 'actual_vols' in realized_data:
        # Get the correct timepoints for the markers
        marker_timepoints = realized_timepoints
        if 'calendar_days' in realized_data:
            marker_timepoints = realized_data['calendar_days']
        
        fig.add_trace(
            go.Scatter3d(
                x=realized_data['actual_moneyness'],
                y=marker_timepoints,
                z=realized_data['actual_vols'],
                mode='markers',
                marker=dict(size=5, color='red'),
                name='Realized Points'
            ),
            row=1, col=1
        )
    
    # 2. Implied Surface
    fig.add_trace(
        go.Surface(
            x=common_K, y=common_T, z=interpolated_options_vol_plot,
            colorscale='Viridis',
            cmin=vmin, cmax=vmax,
            name='Implied Volatility',
            showscale=False
        ),
        row=1, col=2
    )
    
    # Add original option data points
    if 'raw_data' in options_data:
        options_df = options_data['raw_data']
        # Handle the DataFrame correctly
        fig.add_trace(
            go.Scatter3d(
                x=options_df["moneyness"].to_numpy(),
                y=options_df["days"].to_numpy(),
                z=options_df["vol"].to_numpy(),
                mode='markers',
                marker=dict(size=3, color='blue', opacity=0.5),
                name='Option Chain Points'
            ),
            row=1, col=2
        )
    
    # 3. Difference Surface - Red/Blue diverging colorscale
    diff_max = max(abs(np.nanmin(diff_surface)), abs(np.nanmax(diff_surface)))
    diff_min = -diff_max
    
    fig.add_trace(
        go.Surface(
            x=common_K, y=common_T, z=diff_surface_plot,
            colorscale='RdBu_r',  # Red for negative, Blue for positive
            cmin=diff_min, cmax=diff_max,
            name='Volatility Difference',
            showscale=True,
            colorbar=dict(
                title='Diff (Realized - Implied)',
                x=0.45,
                y=0.15,
                len=0.3
            )
        ),
        row=2, col=1
    )
    
    # 4. Slice Comparison (2D for selected maturity)
    # Choose a slice near the middle of the term structure
    slice_idx = realized_T.shape[0] // 2
    slice_days = realized_timepoints[slice_idx] if slice_idx < len(realized_timepoints) else realized_timepoints[-1]
    
    # Extract data for this slice
    slice_moneyness = common_K[slice_idx, :]
    slice_realized = realized_vol[slice_idx, :]
    slice_implied = interpolated_options_vol[slice_idx, :]
    
    # Handle NaNs for plotting
    valid_slice = ~(np.isnan(slice_realized) | np.isnan(slice_implied))
    
    if np.any(valid_slice):
        slice_moneyness = slice_moneyness[valid_slice]
        slice_realized = slice_realized[valid_slice]
        slice_implied = slice_implied[valid_slice]
        
        # Add slice curves
        fig.add_trace(
            go.Scatter(
                x=slice_moneyness,
                y=slice_realized,
                mode='lines+markers',
                line=dict(color='green', width=3),
                name=f'Realized Vol ({slice_days} days)'
            ),
            row=2, col=2
        )
        
        fig.add_trace(
            go.Scatter(
                x=slice_moneyness,
                y=slice_implied,
                mode='lines+markers',
                line=dict(color='purple', width=3),
                name=f'Implied Vol ({slice_days} days)'
            ),
            row=2, col=2
        )
        
        # Add difference
        fig.add_trace(
            go.Scatter(
                x=slice_moneyness,
                y=slice_realized - slice_implied,
                mode='lines',
                line=dict(color='red', width=2, dash='dash'),
                name='Difference'
            ),
            row=2, col=2
        )
        
        # Add zero line
        fig.add_trace(
            go.Scatter(
                x=[min(slice_moneyness), max(slice_moneyness)],
                y=[0, 0],
                mode='lines',
                line=dict(color='black', width=1),
                name='Zero'
            ),
            row=2, col=2
        )
    
    # Update subplot axis properties
    for i in range(1, 3):
        for j in range(1, 3):
            if i < 2 or j < 2:  # 3D plots
                fig.update_scenes(
                    xaxis_title="Log-Moneyness",
                    yaxis_title="Calendar Days",
                    zaxis_title="Annualized Volatility",
                    row=i, col=j
                )
    
    # Update 2D plot
    fig.update_xaxes(title_text="Log-Moneyness", row=2, col=2)
    fig.update_yaxes(title_text="Annualized Volatility", row=2, col=2)
    
    # Create title with key metrics
    stock = realized_data.get('stock', options_data.get('stock', 'Unknown'))
    realized_date = realized_data.get('date', 'Unknown')
    options_date = options_data.get('options_date', 'Unknown')
    
    title = (
        f"Volatility Surface Comparison for {stock}<br>"
        f"Realized: {realized_date} | Options: {options_date}<br>"
        f"RMSE: {metrics.get('rmse', np.nan):.4f} | "
        f"Mean Diff: {metrics.get('mean_diff', np.nan):.4f} | "
        f"Correlation: {metrics.get('correlation', np.nan):.4f}"
    )
    
    # Update layout
    fig.update_layout(
        title=title,
        height=900,
        width=1100,
        margin=dict(l=10, r=10, t=80, b=10)
    )
    
    # Add annotations explaining the meaning of positive/negative differences
    fig.add_annotation(
        xref="paper", yref="paper",
        x=0.25, y=0.35,
        text="Blue = Realized > Implied<br>(Potentially Underpriced Options)",
        showarrow=False,
        font=dict(color="blue", size=12),
        bgcolor="rgba(255,255,255,0.7)",
        bordercolor="blue",
        borderwidth=1,
        borderpad=4
    )
    
    fig.add_annotation(
        xref="paper", yref="paper",
        x=0.25, y=0.30,
        text="Red = Realized < Implied<br>(Potentially Overpriced Options)",
        showarrow=False,
        font=dict(color="red", size=12),
        bgcolor="rgba(255,255,255,0.7)",
        bordercolor="red",
        borderwidth=1,
        borderpad=4
    )
    
    # Pack comparison data
    comparison_data = {
        'metrics': metrics,
        'common_grid': {
            'K': common_K,
            'T': common_T
        },
        'resampled_options_vol': interpolated_options_vol,
        'diff_surface': diff_surface,
        'slice_data': {
            'days': slice_days,
            'moneyness': slice_moneyness,
            'realized': slice_realized,
            'implied': slice_implied
        } if np.any(valid_slice) else None,
        'stock': stock,
        'realized_date': realized_date,
        'options_date': options_date
    }
    
    # Display if requested
    if show_plot:
        fig.show()
    
    return fig, comparison_data

# Compare the surfaces
comparison_fig, comparison_results = compare_surfaces(
    realized_data=ohlcv_surface,
    options_data=options_surface,
    show_plot=True
)

```
